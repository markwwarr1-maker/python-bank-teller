# Bank Teller


In this code I will be implementing code that emulates transactions performed between a bank-teller and a customer. 

The prerequisites are Python 3 syntax, functions and control flow.

### 1. Initializes savings and checking account values. 

```python
checking_balance = 0
savings_balance = 0
```

### 2. Creates a function to check the Balance

##### step 1. Checks account_type and return the respective balance
   
##### step 2. Assigns the savings_balance
   
##### step 3. Assigns the checking_balance
   
##### step 4. Returns an error statement if there are no matching account_type
   
##### step 5. Creates a balance statement
   
##### step 6. Returns balance statement

```python
def check_balance(account_type, checking_balance, savings_balance):
    if account_type == "savings":
        balance = savings_balance
    elif account_type == "checking":
        balance = checking_balance
    else: 
        return "Unsuccessful, please enter \"checking\" or \"savings\""

    balance_statement = "Your " + account_type + " balance is: $" + str(balance)
    return balance_statement
```

### 3. Calling and Printing the check_balance() function for Checking Account

```python
print(check_balance("checking", checking_balance, savings_balance))
```

### 4. Calling and Printing the check_balance() function for Savings Account

```python
print(check_balance("savings", checking_balance, savings_balance))
```

### 5. Creates a function to make a deposit

##### step 1. defines a function
  
##### step 2. Initializes deposit_status variable
   
##### step 3. Ensures deposit is greater than 0 

##### step 4. Error if amount is less than 0
  
##### step 5. Checking account_type
  
##### step 6. Deposits to Savings account
  
##### step 7. Deposits to Checking account
  
##### step 8. Assigns an error statement if there are no matching account_type
  
##### step 9. Creates a deposit statement

##### step 10. Prints deposit statement

##### step 11. Returns savings_balance and checking_balance

```python
def make_deposit(account_type, amount, checking_balance, savings_balance):
  deposit_status = ""
  if amount > 0:
    if account_type == "savings":
      savings_balance += amount
      deposit_status = "successful"
    elif account_type == "checking":
      checking_balance += amount
      deposit_status = "successful"
    else:
      deposit_status = 'Unsuccessful, please enter "checking" or "savings"'
  else:
    deposit_status = "unsuccessful, please enter an amount greater than 0"

  deposit_statement = (
      "Deposit of "
      + str(amount)
      + " to your "
      + account_type
      + " account was "
      + deposit_status
  )
  print(deposit_statement)
  return savings_balance, checking_balance
```

### 6. Call deposit function and make a savings deposit

```python
savings_balance, checking_balance = make_deposit(
    "savings", 10, checking_balance, savings_balance
)
```

### 7. Print savings balance call after making a savings deposit

```python
print(check_balance("savings", checking_balance, savings_balance))
```

### 8. Call deposit function and make a checking deposit

```python
savings_balance, checking_balance = make_deposit(
    "checking", 200, checking_balance, savings_balance
)
```

### 9. Print checking balance call after making a checking deposit

```python
print(check_balance("checking", checking_balance, savings_balance))
```

### 10. Creates a function to make a withdrawal

##### step 1. defines function

##### step 2. Initializes withdrawal_status variable

##### step 3. Initializes an error message

##### step 4. Checking account_type

##### step 5. Withdrawal from savings account

##### step 6. Ensures withdrawal is less than savings account

##### step 7. Subtracts amount from savings account

##### step 8. Withdrawal from checking account

##### step 9. Ensures withdrawal is less than checking account

##### step 10. Subtracts amount from checking account

##### step 11. Assigns an error statement if there are no matching account_type

##### step 12. Creates a withdrawal statement

##### step 13. Prints withdrawal statement

##### step 14. Returns savings_balance and checking_balance

```python
def make_withdrawal(account_type, amount, checking_balance, savings_balance):
    withdrawal_status = ""
    fail = "unsuccessful, please enter amount less than balance"
    if account_type == "savings":
        if amount <= savings_balance:
            savings_balance -= amount
            withdrawal_status = "successful"
        else: withdrawal_status = fail
    elif account_type == "checking":
        if amount <= checking_balance:
            checking_balance -= amount
            withdrawal_status = "successful"
        else: withdrawal_status = fail
    else: withdrawal_status = "unsuccessful, please enter \"checking\" or \"savings\""
    withdrawal_statement = "Withdrawal of $" +str(amount) + " from your " + str(account_type) + " was " + str(withdrawal_status)
    print(withdrawal_statement)
    return (savings_balance, checking_balance)
```

### 11. Calls withdrawal function and make a savings withdrawal

```python
savings_balance, checking_balance = make_withdrawal("savings", 11, checking_balance, savings_balance)
```

### 12. Prints savings balance call, after making a savings withdrawal

```python
print(check_balance("savings", checking_balance, savings_balance))
```

### 13. Calls withdrawal function and make a checking withdrawal

```python
savings_balance, checking_balance = make_withdrawal("checking", 170, checking_balance, savings_balance)
```

### 14. Prints checking balance call, after making a checking withdrawal

```python
print(check_balance("checking", checking_balance, savings_balance))
```

### 15. Creates a function to make a transfer between accounts

##### step 1. defines a function

##### step 2. Initializes transaction_status variable

##### step 3. Initializes an error message

##### step 4. Account Transfer

##### step 5. Ensure transfer is less than savings account

##### step 6. Transfer amount from savings to checking account

##### step 7. Ensure transfer is less than checking account

##### step 8. Transfer amount from checking to savings  account

##### step 9. Assign an error statement if there are no matching account_type

##### step 10. Create a transfer statement

##### step 11. Print transfer statement

##### step 12. Return savings_balance and checking_balance

```python
def acc_transfer(acc_from, acc_to, amount, checking_balance, savings_balance):
    transaction_status = ""
    trans_error = "unsuccessful, please enter amount less than "
    if acc_from == "savings" and acc_to == "checking":
        if amount <= savings_balance:
            savings_balance -= amount
            checking_balance += amount
            transaction_status = "successful"
        else: 
            transaction_status = trans_error + str(savings_balance)
    elif acc_from == "checking" and acc_to == "savings":
        if checking_balance >= amount:
            checking_balance -= amount
            savings_balance += amount
            transaction_status = "successful"
        else: 
            transaction_status = trans_error +str(checking_balance)
    else: 
        transaction_status = "unsuccessful, please enter \"checking\" or \"savings\""
    transaction_statement = "Transfer of " + str(amount) + " from your " + str(acc_from) + " to your " + str(acc_to) + " account was " + str(transaction_status)
    print(transaction_statement)
    return (savings_balance, checking_balance)
```

### 16. Calls transfer function and make a checking to savings transfer

```python
(savings_balance, checking_balance) = acc_transfer("checking", "savings", 40, checking_balance, savings_balance)
```

### 17. Prints checking balance after making a checking to savings transfer

```python
print(check_balance("checking", checking_balance, savings_balance))
```

### 18. Prints savings balance after making a checking to savings transfer

```python
print(check_balance("savings", checking_balance, savings_balance))
```
